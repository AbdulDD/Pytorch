```python
from torch.utils.data import TensorDataset, DataLoader

import torch
import torchvision
import torch.nn as nn
```

# Load Dataset

```python
import tensorflow as tf


# Load the MNIST dataset directly from keras datasets
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

# Pytorch Tensor conversion and Data Loader

```python
# Convert TensorFlow tensors to Pytorch Tensors

x_train = torch.tensor(x_train, dtype=torch.float32)
x_test = torch.tensor(x_test, dtype=torch.float32)

y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)

```

```python
# Check the dataset
print('------------------------------------------------')
print('Training dataset')
print(f'Shape of image samples: {x_train.shape}')
print(f'Shape of target samples: {y_train.shape}')
print('------------------------------------------------')
print('Testing dataset')
print(f'Shape of image samples: {x_test.shape}')
print(f'Shape of target samples: {y_test.shape}')
```

```python
x_train = x_train.unsqueeze(1)
x_test = x_test.unsqueeze(1)
```

```python
print(f'Training Images Shape: {x_train.shape}, Testing Images Shape: {x_test.shape}')
```

```python
# Create Tensor Dataset
train_dataset = TensorDataset(x_train, y_train)
test_dataset = TensorDataset(x_test, y_test)
```

```python
image, label = train_dataset[7]
print(f'Tensor shape of image: {image.shape}, Tensor shape of label: {label.shape}')
```

```python
# DataLoader
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)
```

```python
for image, label in train_loader:
  print(label)
  break
```

# Architecture Design

```python
# Convolution Layer
CovLayer1 = nn.Conv2d(1, 5, 5)

#Pooling Layer
Poollayer1 = nn.MaxPool2d(2,2)

# ReLu
ReLU1 = nn.ReLU()

# Convolution Layer
CovLayer2 = nn.Conv2d(5, 16, 5)

# ReLu
ReLU2 = nn.ReLU()

#Pooling Layer
Poollayer2 = nn.MaxPool2d(2,2)


# Flatten Layer
Flattenlayer = nn.Flatten()

# Dense Layer
DenseLayer1 = nn.Linear(256, 64)

# ReLU3
ReLU3 = nn.ReLU()

# Dense Layer
DenseLayer2 = nn.Linear(64, 10)


```

```python
# Hyper parameters

# Loss Function
loss_fn = nn.CrossEntropyLoss()

# Epochs & Learning rate
Epochs = 10
lr = 0.01

# Optimizer
optimizer = torch.optim.Adam(
    [
        *CovLayer1.parameters(),
        *CovLayer2.parameters(),
        *DenseLayer1.parameters(),
        *DenseLayer2.parameters(),

    ],
    lr = lr
)

```

# Training Custom LeNet-5 on MNIST Dataset

```python
# Training Loop

for epoch in range(Epochs):

  total_loss = 0

  for image, label in train_loader:


    # Forward Pass start
    # Convolution Block 1
    x = CovLayer1(image)
    x = ReLU1(x)
    x = Poollayer1(x)

    # Convolution Block 2
    x = CovLayer2(x)
    x = ReLU2(x)
    x = Poollayer2(x)

    # Flatten Layer
    x = Flattenlayer(x)

    # Dense Layer
    x = DenseLayer1(x)
    x = ReLU3(x)

    # Output Layer (Provides Logits)
    Logit = DenseLayer2(x)

    # Forward Pass Ends

    # Calculate Loss
    Loss = loss_fn(Logit, label)

    # Gradients to zero
    optimizer.zero_grad()

    # Backpropogation
    Loss.backward()

    # Update Weights
    optimizer.step()

    # Add batch loss
    total_loss += Loss.item()

  # Calculate epoch loss
  epoch_loss = total_loss/len(train_loader)

  # Print epoch loss
  print(f'Epoch: {epoch + 1}/{Epochs} ----> Loss: {epoch_loss:.4f}')







```
